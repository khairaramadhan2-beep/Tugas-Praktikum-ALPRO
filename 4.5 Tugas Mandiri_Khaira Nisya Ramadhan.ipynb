```python
# Input data dari pengguna
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam rupiah): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").lower()

# 1. Menentukan tingkat risiko awal berdasarkan Jenis Kendaraan (Eksplisit)
if jenis_kendaraan == "motor":
    indeks_risiko = 1  # Sedang
elif jenis_kendaraan == "mobil":
    indeks_risiko = 0  # Rendah
else:
    indeks_risiko = 0  # Default jika input tidak sesuai

# 2. Faktor Usia Pengemudi (Terlalu muda < 20 ATAU terlalu tua > 60 -> Otomatis Risiko Tinggi)
if usia_pengemudi < 20 or usia_pengemudi > 60:
    indeks_risiko = 2  # Langsung ke level tertinggi (Tinggi)

# 3. Faktor Area Penggunaan (Perkotaan -> Risiko meningkat 1 level)
if area_penggunaan == "perkotaan":
    if indeks_risiko < 2:
        indeks_risiko += 1

# 4. Faktor Sistem Pengaman (Tidak ada pengaman -> Risiko meningkat 1 level)
if sistem_pengaman == "tidak":
    if indeks_risiko < 2:
        indeks_risiko += 1

# Konversi indeks angka menjadi teks kategori risiko
tingkat_risiko = ["rendah", "sedang", "tinggi"]
kategori_risiko = tingkat_risiko[indeks_risiko]

# 5. Menentukan persentase premi berdasarkan Kategori Risiko Akhir (if-elif-else)
if kategori_risiko == "tinggi":
    persentase_premi = 0.1
elif kategori_risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

# Hitung Total Premi
premi_dibayarkan = persentase_premi * nilai_kendaraan

# 6. Pesan Doraemon menggunakan Conditional Expression
pesan = "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!" if kategori_risiko == "tinggi" else "Risiko kendaraan masih aman menurut Doraemon"

# Menampilkan Hasil Akhir
print("=== HASIL ANALISIS ASURANSI ===")
print(f"Jenis Kendaraan       : {jenis_kendaraan.capitalize()}")
print(f"Kategori Risiko Akhir : {kategori_risiko.upper()}")
print(f"Persentase Premi      : {persentase_premi * 100}%")
print(f"Total Premi Bulanan   : Rp {premi_dibayarkan:,.2f}")
print(f"Pesan Doraemon        : {pesan}")
```
