```python
# a
hari = 1
saldo = 50000
while hari <= 30:
    pengeluaran = int(input(f"Pengeluaran hari ke-{hari}: Rp. "))
    sisa_saldo = saldo - pengeluaran
    saldo = sisa_saldo
    if sisa_saldo <= 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan pengeluaran.")
        break
    else:
        print(f"Sisa saldo hari ke-{hari} sebesar Rp. {saldo}")
        hari += 1 # untuk pengulangan 
```

```python
# b
# Menganalisis pengeluaran mingguan Nobita
minggu = 0
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in pengeluaran_mingguan:
    minggu += 1
    if i == 0:
       print(f"Minggu ke-{minggu} Nobita tidak jajan, dilewati analisisnya. ")
    elif i != 0:
       print(f"Minggu ke-{minggu} Total pengeluaran Rp. {i}")
```

```python
# c
bulan = [1, 2, 3]
minggu = [1, 2, 3, 4]

for i in bulan:
    total_bulanan = 0
    print(f"\nBulan ke-{i}")
    
    pencatatan_berjalan = True
    
    for r in minggu:
        pengeluaran = int(input(f"Pengeluaran minggu ke-{r}: Rp. "))
        
        if pengeluaran > 20000:
            print("Total pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.\n")
            total_bulanan = total_bulanan + pengeluaran
            pencatatan_berjalan = False
            break
        else:
            total_bulanan = total_bulanan + pengeluaran
            
    print(f"Total pengeluaran bulan ke-{i}: Rp. {total_bulanan}\n")
```

```python
bulan = [1, 13]
saldo =  300000
saldo_awal = saldo

print(f"\nEvaluasi Akhir Tahun")
for bulan in range(1,13):
    print(f"===== Bulan ke-{bulan} =====")
    minggu = 1
    while minggu <= 4:
        pertanyaan = input(f"Apakah minggu ke-{minggu} libur? y/n: ")
        minggu += 1
        if pertanyaan == "n":
            pengeluaran = int(input("Masukkan pengeluaran minggu ini: RP. "))
            saldo -= pengeluaran
            if saldo <= 0:
                break
    if saldo <= 0:
        print("Simulasi berhenti! (break)")
        break
    else:
        print(f"Sisa saldo tersisa akhir bulan{bulan} Rp.{saldo}\n\n")

if saldo > 0:
    print(f"===== Nobita berhasil menabung Rp.{saldo} dari total Rp.{saldo_awal}! =====")
```
