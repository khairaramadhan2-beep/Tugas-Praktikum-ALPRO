```python
%%writefile interest.py

def compound_interest(principal, rate, time):
    """hitung nilai akhir dengan bunga majemuk"""
    return principal * (1 + rate)**time

def simple_interest(principal, rate, time):
    """hitung nilai akhir dengan bunga sederhana"""
    return principal * (1 + rate * time)
```

```python
%%writefile display.py

def print_result(principal, rate, years, ci, si):
    print("===== HASIL PERHITUNGAN TABUNGAN =====")
    print(f"Saldo Awal        : ¥{principal:,.0f}")
    print(f"Bunga Tahunan     : {rate*100:.1f}%")
    print(f"Lama Menabung     : {years} tahun")
    print("---------------------------------------")
    print(f"Nilai Akhir (CI)  : ¥{ci:,.0f}")
    print(f"Nilai Akhir (SI)  : ¥{si:,.0f}")
    print("=======================================")
```

```python
%%writefile main.py

import interest
import display

print("=== Program Kalkulator Bunga Tabungan ===")

principal = float(input("Masukkan Saldo Awal       : ¥"))
rate = float(input("Masukkan Bunga Tahunan (%) : ")) / 100
years = int(input("Masukkan Lama Menabung (tahun): "))

ci = interest.compound_interest(principal, rate, years)
si = interest.simple_interest(principal, rate, years)

display.print_result(principal, rate, years, ci, si)
```
