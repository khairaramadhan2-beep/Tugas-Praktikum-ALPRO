```python
# ================================
# Variabel Global & Fungsi Update 
# ================================
skor_nobita = 0

def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai

def belajar(jam_belajar):
    if jam_belajar <= 1:
        nilai = -10
    elif 1 < jam_belajar < 3:
        nilai = 20
    else:
        nilai = 15
    update_skor(nilai)

def tidur(jam_tidur):
    if jam_tidur <= 5:
        nilai = -10
    elif 5 < jam_tidur <= 8:
        nilai = 20
    else:
        nilai = -15
    update_skor(nilai)

def main_game(opsi_screen_time, jumlah_gadget, waktu_main):
    if opsi_screen_time == 1 and jumlah_gadget >= 2 and waktu_main > 3:
        nilai = -30
    elif opsi_screen_time == 2 and jumlah_gadget < 2 and waktu_main <= 2:
        nilai = 15
    else:
        nilai = 0
    update_skor(nilai)

# ====================
# 4. Fungsi Menyontek
# ====================
def menyontek():
    # Fungsi tidak membutuhkan parameter, namun mengembalikan/mempunyai nilai kembalian
    nilai = -50
    print("Nobita ketahuan menyontek!")
    update_skor(nilai)

# ==========================================
# d. Membuat Fungsi Pengendali Aktivitas
# ==========================================
def pengendali_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))
        belajar(jam_belajar)
    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam Nobita tidur hari ini? "))
        tidur(jam_tidur)
    elif pilihan == 3:
        print("Kondisi gadget Nobita:")
        print("1. Layar menyala terus / Boros screen time")
        print("2. Penggunaan bijak")
        opsi = int(input("Pilih opsi screen time (1/2): "))
        jml_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu = float(input("Berapa jam Nobita bermain gadget? "))
        main_game(opsi, jml_gadget, waktu)
    elif pilihan == 4:
        menyontek()
    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan.")

# =================================
# e. Membuat Fungsi Evaluasi Akhir
# =================================
def evaluasi_akhir():
    if skor_nobita >= 80:
        return "Siswa Teladan"
    elif 50 <= skor_nobita < 80:
        return "Cukup Baik"
    else: # skor_nobita < 50
        return "Pemalas"

# ==============================
# f. Membuat Menu Utama Program
# ==============================
# ==========================================
# f. Membuat Menu Utama Program (Perbaikan)
# ==========================================
def menu_utama():
    while True:
        print("\n=== PROGRAM PERINGKAT DISIPLIN NOBITA ===")
        print("1. Belajar")
        print("2. Tidur")
        print("3. Main bersama Doraemon")
        print("4. Menyontek saat ujian")
        print("5. Keluar & Lihat Evaluasi Akhir")
        print("=========================================")
        
        # Menggunakan try-except untuk menghindari error jika user tidak sengaja mengetik huruf
        try:
            pilihan = int(input("Pilih aktivitas (1-5): "))
        except ValueError:
            print("Masukkan harus berupa angka 1 sampai 5!")
            continue
        
        # Validasi pilihan 5 diproses paling atas agar program langsung keluar dengan benar
        if pilihan == 5:
            status = evaluasi_akhir()
            print(f"\nSkor Akhir Nobita: {skor_nobita}")
            print(f"Hasil Evaluasi: {status}")
            print("Program selesai. Terima kasih!")
            break
        elif 1 <= pilihan <= 4:
            pengendali_aktivitas(pilihan)
        else:
            print("Pilihan tidak valid! Silakan pilih menu antara 1 sampai 5.")

# Menjalankan Program Utama
if __name__ == "__main__":
    menu_utama()
```
