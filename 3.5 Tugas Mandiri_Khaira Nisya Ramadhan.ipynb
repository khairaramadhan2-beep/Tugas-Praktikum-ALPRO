```python
# Membuat list belanjaan menggunakan input
barang_1 = input("Masukkan barang pertama: ")
barang_2 = input("Masukkan barang kedua: ")
barang_3 = input("Masukkan barang ketiga: ")

belanjaan = [barang1, barang2, barang3]
print("Daftar belanjaan:", ", ".join(belanjaan))

# Membuat tuple harga dan jumlah
harga_jumlah = 5000, 3, 12000, 2, 3000, 1
print("Harga dan jumlah:", harga_jumlah)

# Menampilkan karakter pertama dari barang pertama
print("Karakter pertama dari barang pertama:", belanjaan[0][0])

# Menampilkan karakter terakhir dari barang terakhir
print("Karakter terakhir dari barang terakhir:", belanjaan[-1][-1])
```

```python
# Menghitung total harga masing-masing barang
total_apel = 5000 * 3
total_susu = 12000 * 2
total_roti = 3000 * 1

# Menampilkan total harga masing-masing barang
print("Total harga Apel:", total_apel)
print("Total harga Susu:", total_susu)
print("Total harga Roti:", total_roti)

# Menghitung total keseluruhan belanja
total_belanja = total_apel + total_susu + total_roti

# Menampilkan total keseluruhan belanja
print("Total keseluruhan belanja:", total_belanja)
```
