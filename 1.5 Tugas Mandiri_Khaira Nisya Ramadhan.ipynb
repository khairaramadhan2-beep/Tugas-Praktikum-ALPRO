```python
#Menyimpan modal Doraemon
modal_doraemon = 555000

#Menyimpan modal Nobita
modal_nobita = 333000

#Menyimpan modal Shizuka
modal_shizuka = 222000
```

```python
#Menghitung total modal
total_modal = modal_doraemon + modal_nobita + modal_shizuka

#Menampilkan total modal
print("Total modal adalah Rp", total_modal)
```

```python
#Menyimpan jumlah dorayaki yang terjual
jumlah_dorayaki = 350

#Menyimpan harga satu dorayaki
harga_dorayaki = 5000

#Menghitung total pendapatan
total_pendapatan = jumlah_dorayaki * harga_dorayaki

#Menampilkan total pendapatan
print("Total pendapatan adalah Rp", total_pendapatan)
```

```python
#Menghitung keuntungan
keuntungan = total_pendapatan - total_modal

#Menampilkan keuntungan
print("Keuntungan total adalah Rp", keuntungan)
```
